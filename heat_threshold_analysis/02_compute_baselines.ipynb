# 02 — Compute Baselines

Reads the cached NetCDF files written by `01_data_acquisition` and computes
**per-pixel historical Tmax percentiles** for each city over the full 2003–2020
record.  Results are written to `data/{city}_baseline.nc`.

**Re-run this notebook if `01_data_acquisition` is re-run** (e.g. date range
changed or a city is added).  It does not contact GEE.

Expected runtime: seconds to a few minutes per city (pure NumPy/xarray).

## What this notebook computes

For each city and each grid pixel, `percentile_baseline` calls
`xr.DataArray.quantile` over the entire time axis.  The result is a
DataArray with dimensions `(percentile, lat, lon)` where the `percentile`
coordinate holds the integer labels `[50, 75, 90, 95, 99]`.

The 90th-percentile layer is the default relative threshold used in
`03_threshold_analysis` for the spatial maps and neighbourhood quadrant plots.
All five layers are saved so you can experiment with other percentiles without
recomputing.

```python
import os
import sys
import xarray as xr

sys.path.insert(0, os.path.dirname(os.path.abspath('__file__')))
from utils.heat import CITIES, load_city, percentile_baseline
```

## Configuration

```python
DATA_DIR    = 'data'
PERCENTILES = [50, 75, 90, 95, 99]

print(f'Percentiles to compute : {PERCENTILES}')
print(f'Data directory         : {DATA_DIR}/')
```

## Compute and save per-pixel baselines

Each cell below processes one city independently.

### Salvador

```python
da_sal      = load_city('salvador', DATA_DIR)
bl_sal      = percentile_baseline(da_sal, PERCENTILES)
out_sal     = f'{DATA_DIR}/salvador_baseline.nc'
bl_sal.to_dataset(name='tmax').to_netcdf(out_sal)
print(f'Saved: {out_sal}')
print(bl_sal)
```

### Teresina

```python
da_ter      = load_city('teresina', DATA_DIR)
bl_ter      = percentile_baseline(da_ter, PERCENTILES)
out_ter     = f'{DATA_DIR}/teresina_baseline.nc'
bl_ter.to_dataset(name='tmax').to_netcdf(out_ter)
print(f'Saved: {out_ter}')
print(bl_ter)
```

### Cáceres

```python
da_cac      = load_city('caceres', DATA_DIR)
bl_cac      = percentile_baseline(da_cac, PERCENTILES)
out_cac     = f'{DATA_DIR}/caceres_baseline.nc'
bl_cac.to_dataset(name='tmax').to_netcdf(out_cac)
print(f'Saved: {out_cac}')
print(bl_cac)
```

## Verify outputs

For each city, print the spatial-mean temperature at each percentile level.
These values should increase monotonically and fall within a plausible range
for the city's climate.

```python
import pandas as pd

records = []
for city in ['salvador', 'teresina', 'caceres']:
    bl = xr.open_dataset(f'{DATA_DIR}/{city}_baseline.nc')['tmax']
    for pct in PERCENTILES:
        val = float(bl.sel(percentile=pct).mean())
        records.append({'city': city, 'percentile': pct, 'tmax_mean_C': round(val, 2)})

df = pd.DataFrame(records)
print(df.pivot(index='city', columns='percentile', values='tmax_mean_C').to_string())
```

## TX90p calendar-day threshold

Computes the **ETCCDI TX90p** threshold for each city using
`xclim.core.calendar.percentile_doy` with a centred 5-day window.
Unlike the flat percentile in the section above (one scalar per pixel),
this produces one threshold value per **(dayofyear, lat, lon)** cell,
so each calendar day is judged against its own seasonal norm.

Results are saved to `data/{city}_tx90p_thresh.nc`.

```python
from utils.heat import compute_tx90p_threshold, load_tx90p_threshold

# P99 excluded: 18 years x 5-day window = ~90 samples per calendar day per pixel.
# P99 of 90 samples is essentially the single highest observed value, making it
# highly sensitive to outliers and measurement artifacts in the GSHTD source data.
TX90P_PERCENTILES = [90, 95]

for city in ['salvador', 'teresina', 'caceres']:
    da = load_city(city, DATA_DIR)
    da.attrs['units'] = 'degC'

    thresh = compute_tx90p_threshold(da, window=5, per=TX90P_PERCENTILES)

    out = f'{DATA_DIR}/{city}_tx90p_thresh.nc'
    thresh.to_dataset(name='tmax').to_netcdf(out)
    print(f'Saved: {out}  shape {dict(zip(thresh.dims, thresh.shape))}')
```

```python
# Diagnostic: confirm thresholds vary across dayofyear for each percentile.
# Near-zero std would indicate the window had no effect.
print('City-wide mean TX threshold — std across dayofyear:')
print('(near-zero would indicate the window had no effect)\n')
for city in ['salvador', 'teresina', 'caceres']:
    thresh = load_tx90p_threshold(city, DATA_DIR)
    for p in TX90P_PERCENTILES:
        spatial_mean = thresh.sel(percentile=p).mean(dim=['lat', 'lon'])
        std_doy = float(spatial_mean.std())
        t_min   = float(spatial_mean.min())
        t_max   = float(spatial_mean.max())
        print(f'  {city:10s} P{p}: std={std_doy:.2f} °C   range {t_min:.1f}–{t_max:.1f} °C')
```

## TN90p calendar-day threshold

Computes the **ETCCDI TN90p** threshold for each city — identical method to TX90p above but applied to daily minimum temperature (Tmin).  Results are saved to `data/{city}_tn90p_thresh.nc`.

**Prerequisites:** `01b_data_acquisition_tmin` must have been run first.

```python
from utils.heat import load_city_tmin

TN90P_PERCENTILES = [90, 95]

for city in ['salvador', 'teresina', 'caceres']:
    da = load_city_tmin(city, DATA_DIR)
    da.attrs['units'] = 'degC'

    thresh = compute_tx90p_threshold(da, window=5, per=TN90P_PERCENTILES)

    out = f'{DATA_DIR}/{city}_tn90p_thresh.nc'
    thresh.to_dataset(name='tmin').to_netcdf(out)
    print(f'Saved: {out}  shape {dict(zip(thresh.dims, thresh.shape))}')
```

## Population grids (1 km)

Resamples the 100m WorldPop rasters onto the 1km GSHTD temperature grid for
each city and year, then caches the result to `data/{city}_pop_1km_{year}.nc`.

This is the slow step (point-in-cell aggregation for 18 years × 3 cities).
Downstream notebooks read from cache and never recompute.

**Prerequisites:** `01_data_acquisition` population section must be run first.

```python
import os, sys
import xarray as xr
import numpy as np
from pathlib import Path

sys.path.insert(0, os.path.dirname(os.path.abspath('__file__')))
from utils.heat import CITIES, load_city, load_population, resample_population_to_grid

DATA_DIR   = 'data'
POP_YEARS  = list(range(2003, 2021))
CITY_KEYS  = ['salvador', 'teresina', 'caceres']
```

```python
for city in CITY_KEYS:
    label  = CITIES[city]['label']
    da_tmp = load_city(city, DATA_DIR)       # 1km temperature grid (template)
    pop_da = load_population(city, DATA_DIR) # 100m population (year, lat, lon)

    print(f'\n{label}')
    for yr in POP_YEARS:
        out = Path(DATA_DIR) / f'{city}_pop_1km_{yr}.nc'
        if out.exists():
            print(f'  {yr}: cache exists — skipping')
            continue

        # Select population for this year (fall back to nearest if missing)
        avail_years = pop_da.year.values.tolist()
        yr_use = yr if yr in avail_years else min(avail_years, key=lambda y: abs(y - yr))
        pop_yr = pop_da.sel(year=yr_use)

        pop_1km = resample_population_to_grid(pop_yr, da_tmp.isel(time=0))
        pop_1km.to_dataset().to_netcdf(out)
        print(f'  {yr}: saved  (proxy year={yr_use}, '
              f'total pop={float(pop_1km.sum()):,.0f})')

print('\nDone.')
```

### Verify — total city population by year

```python
import pandas as pd

rows = []
for city in CITY_KEYS:
    for yr in [2003, 2010, 2015, 2020]:
        f = Path(DATA_DIR) / f'{city}_pop_1km_{yr}.nc'
        if f.exists():
            pop = xr.open_dataset(f)['population']
            rows.append({'city': CITIES[city]['label'], 'year': yr,
                         'total_pop': int(pop.sum())})

df = pd.DataFrame(rows)
print(df.pivot(index='city', columns='year', values='total_pop')
        .applymap(lambda x: f'{x:,}').to_string())
```

## Population grids (100 m, per year — Salvador only)

Slices the stacked `salvador_population.nc` (100m WorldPop, geographic lat/lon)
into individual per-year files `data/salvador_pop_100m_{year}.nc`.

These are kept in the native WorldPop CRS (EPSG:4326) and are intended for use
with the UrbClim 100m temperature data in `08_urbclim_100m_salvador`, where they
are reprojected onto the UrbClim UTM grid at runtime.

The 1km per-year files produced above are kept unchanged.

```python
pop_sal_100m = load_population('salvador', DATA_DIR)  # (year, lat, lon) at 100m

for yr in POP_YEARS:
    out = Path(DATA_DIR) / f'salvador_pop_100m_{yr}.nc'
    if out.exists():
        print(f'{yr}: cache exists — skipping')
        continue
    pop_yr = pop_sal_100m.sel(year=yr).drop_vars('year')
    pop_yr.to_dataset().to_netcdf(out)
    print(f'{yr}: saved  (total pop = {float(pop_yr.sum()):,.0f})')

print('\nDone — per-year 100m files written to data/salvador_pop_100m_YYYY.nc')
```
